```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('flight_wifi_data.csv')
```

```python
data.head()
```

```python
data["UNIQUE_FLIGHT_ID"] = (
    data['flight_number'].astype(str).str.strip()
    + " | "
    + data["departure_date"].astype(str)
    + " | "
    + data["origin"].astype(str).str.strip()
    + "-"
    + data["destination"].astype(str).str.strip()
)
```

```python
data['UNIQUE_FLIGHT_ID'].value_counts()
```

```python
nps_data = pd.read_csv('nps_wifi_score_all2026.csv')
```

```python
nps_data.head()
```

```python
nps_data["UNIQUE_FLIGHT_ID"] = (
    nps_data['OPERAT_FLIGHT_NBR'].astype(str).str.strip()
    + " | "
    + nps_data["SEG_DEP_DT"].astype(str)
    + " | "
    + nps_data["SEG_DEP_AIRPRT_IATA_CD"].astype(str).str.strip()
    + "-"
    + nps_data["SEG_ARVL_AIRPRT_IATA_CD"].astype(str).str.strip()
)
```

```python
nps_data["UNIQUE_FLIGHT_ID"]
```

```python
data_combined = data.merge(nps_data, how = 'inner', on = 'UNIQUE_FLIGHT_ID')
```

```python
pd.set_option('display.max_columns', None)
data_combined.head()
```

```python
# Convert to datetime
data_combined['actual_arrival_time'] = pd.to_datetime(
    data_combined['actual_arrival_time'],
    errors='coerce'
)

data_combined['scheduled_arrival_time'] = pd.to_datetime(
    data_combined['scheduled_arrival_time'],
    errors='coerce'
)

# Calculate delay
# Result will automatically be NaT if either timestamp is missing
data_combined['delay'] = (
    data_combined['actual_arrival_time']
    - data_combined['scheduled_arrival_time']
)

data_combined['delay_mins'] = (
    data_combined['actual_arrival_time']
    - data_combined['scheduled_arrival_time']
).dt.total_seconds() / 60

data_combined['delay_bucket'] = pd.cut(
    data_combined['delay_mins'],
    bins=[-np.inf, 0, 120, 240, 720, np.inf],
    labels=['Early', '0-2 hrs', '2-4 hrs', '4-12 hrs', '>12 hrs']
)
```

```python
data_combined1 = data_combined[['UNIQUE_FLIGHT_ID', 'flight_number',  'origin', 'destination', 'departure_date',
       'tail_number', 'equipment_type', 'pax_count', 'wifi_sessions', 'delay_mins', 'delay_bucket','SURVEY_ID', 'SURVEY_RESPNS_ID',
       'SURVEY_AA_CUST_UNIQUE_ID', 'SURVEY_LYLTY_ACCT_ID','FLEET_CD',
       'SUBFLEET_CD', 'CABIN_FLOWN_CD', 'WIFI_SATISFACTION_SCR',
       'WIFI_SPEED_SCR', 'WIFI_PRICE_SCR', 'WIFI_RLBLTY_SCR',
       'WIFI_EASE_USE_SCR', 'WIFI_CNCT_EFFRT_SCR', 'LIKELIHOOD_RECOMMEND_SCL',
       'NET_PROMO_CATG']]
```

```python
flight_variation = (
    data_combined1.groupby('UNIQUE_FLIGHT_ID')['LIKELIHOOD_RECOMMEND_SCL']
    .agg(
        Count='count',
        Min='min',
        Max='max'
    )
    .reset_index()
)

flight_variation.sort_values(by = 'Count', inplace = True, ascending = False)
flight_variation.head()
```
